````python
# PART 1: Measures of Central Tendency

# 1.1 Mean (Average)

## Batters Dataset Questions

# 1. Find the average (mean) runs scored by all batters

batters_avg = df_batters["Runs"].mean()

print(f"📊 Mean runs scored by all batters: {batters_avg:.2f}")

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(df_batters["Runs"], alpha=0.7, edgecolor="black")

# Mean Line
plt.axvline(
    batters_avg,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {batters_avg:.2f}",
)

# Labels and Title
plt.xlabel("Runs Scored")
plt.ylabel("Frequency")
plt.title("Distribution of Runs Scored by Batters")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# Interpretation
print(
    f"\n💡 Interpretation: "
    f"The average runs scored by batters in IPL 2025 is approximately {batters_avg:.2f}."
)
````

````python
# 2. Calculate mean strike rate (SR) of all batters.
batters_sr = df_batters["SR"].mean()
print(f"\n📊 Mean Strike Rate by all batters: {batters_sr:.2f}")

# Visualisation
plt.figure(figsize=(10, 5))
plt.hist(df_batters["SR"], edgecolor="black")
plt.axvline(
    batters_sr,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean Strike Rate = {batters_sr:.2f}",
)

plt.xlabel("Strike Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Strike Rate for All Batters")
plt.grid(True, alpha=0.3)

plt.legend()
plt.show()

print(
    f"\n💡 Interpretation : The average strike rate for Batters in IPL 2025 was approximately {batters_sr:.2f}"
)
````

````python
# 3. Find average number of sixes hit by batters.
sixes_hit = df_batters["6s"].mean()
print(f"📊 Mean Sixes hit by all batters : {sixes_hit:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_batters["6s"], edgecolor="black", alpha=0.7)
plt.axvline(
    sixes_hit, linestyle="--", linewidth=2, label=f"Mean sixes hit {sixes_hit:.2f}"
)
plt.legend()
plt.xlabel("Sixes Hit By Batter")
plt.ylabel("Frequency")
plt.title("Distribution of Sixes Hit By Batters")
plt.grid(alpha=0.3)
plt.show()

print(
    f"💡 Interpretation : The average Sixes Hit by the Batters in IPL 2025 is {sixes_hit:.2f}"
)
````

````python
# 4. Which batter scored closest to the average runs?
# Average Runs
average_runs = df_batters["Runs"].mean()

# Closest Batter Index
closest_index = (df_batters["Runs"] - average_runs).abs().idxmin()

# Closest Batter Data
closest_batter = df_batters.loc[closest_index]

# Values
closest_runs = closest_batter["Runs"]
closest_name = closest_batter["Player Name"]

# Plot
plt.figure(figsize=(10, 5))

# Histogram
plt.hist(df_batters["Runs"], bins=10, edgecolor="black", alpha=0.7)

# Mean Line
plt.axvline(
    average_runs,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean Runs = {average_runs:.2f}",
)

# Closest Batter Line
plt.axvline(
    closest_runs,
    color="green",
    linestyle="-",
    linewidth=2,
    label=f"{closest_name} = {closest_runs}",
)

# Labels
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("Runs Distribution with Closest Batter to Average")

plt.legend()

plt.grid(True, alpha=0.3)
plt.show()
````

````python
# 5. Find mean batting average (AVG) team-wise.
# Team Color Mapping
team_colors = {
    "MI": "#004BA0",
    "CSK": "#F9CD05",
    "RCB": "#D71920",
    "KKR": "#3A225D",
    "SRH": "#FF822A",
    "DC": "#17449B",
    "RR": "#EA1A85",
    "PBKS": "#D71920",
    "LSG": "#00AEEF",
    "GT": "#1B2133",
}


# Convert AVG column to numeric
df_batters["AVG"] = pd.to_numeric(df_batters["AVG"], errors="coerce")

# Team-wise batting average
team_wise_avg = df_batters.groupby("Team")["AVG"].mean().sort_values(ascending=False)

print("\n📊 Team-wise Mean Batting Average:\n")
print(team_wise_avg)

colors = [team_colors[team] for team in team_wise_avg.index]

# Visualization
plt.figure(figsize=(10, 5))

ax = team_wise_avg.plot(kind="bar", edgecolor="black", alpha=0.7, color=colors)

plt.xlabel("Team")
plt.ylabel("Mean Batting Average")
plt.title("Team-wise Mean Batting Average")
plt.bar_label(ax.containers[0], fmt="%.2f")

plt.grid(True, alpha=0.3)

plt.show()

# Interpretation
print(
    "\n💡 Interpretation: "
    "This graph shows the average batting performance of players across different IPL teams."
)
````

````python
# 4. Compare mean vs median runs. Is the data skewed?

# Median
median_runs = df_batters["Runs"].median()
print(f"📊 Median Runs Scored: {median_runs:.2f}")

# Mean
mean_runs = df_batters["Runs"].mean()
print(f"📊 Mean Runs Scored: {mean_runs:.2f}")

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(df_batters["Runs"], bins=10, alpha=0.7, edgecolor="black")

# Median Line
plt.axvline(
    median_runs,
    linestyle="--",
    linewidth=2,
    color="blue",
    label=f"Median = {median_runs:.2f}",
)

# Mean Line
plt.axvline(
    mean_runs, linestyle="-", linewidth=2, color="red", label=f"Mean = {mean_runs:.2f}"
)

# Labels and Title
plt.xlabel("Runs Scored")
plt.ylabel("Frequency")
plt.title("Distribution of Runs Scored")

plt.grid(True, alpha=0.3)

plt.legend()

plt.show()

# Skewness Interpretation
if mean_runs > median_runs:
    print(
        "\n💡 Interpretation: "
        "The data is right skewed because the mean is greater than the median. "
        "This indicates the presence of a few high-scoring batters pulling the average upward."
    )

elif mean_runs < median_runs:
    print(
        "\n💡 Interpretation: "
        "The data is left skewed because the mean is smaller than the median."
    )

else:
    print(
        "\n💡 Interpretation: "
        "The data is approximately symmetric because the mean and median are nearly equal."
    )
````

````python
# 1.3 Mode

## Batters

# 1. Find mode of matches played.
matches_played_mode = df_batters["Matches"].mode()[0]
print(f"\n📊 Mode of Matches Played by batters: {matches_played_mode}")

plt.figure(figsize=(12, 5))

ax = sns.countplot(x=df_batters["Matches"])

# Get category positions
categories = sorted(df_batters["Matches"].unique())

# Find mode position
mode_position = categories.index(matches_played_mode)

# Highlight only mode bar
for bar, label in zip(ax.patches, sorted(df_batters["Matches"].unique())):

    if label == matches_played_mode:
        bar.set_color("red")
    else:
        bar.set_color("skyblue")


plt.text(
    mode_position,
    ax.patches[mode_position].get_height() + 1,
    f"Mode = {matches_played_mode}",
    ha="center",
    color="red",
)

plt.xlabel("Matches Played")
plt.ylabel("Frequency")
plt.title("Mode of Matches Played")

plt.show()

print(
    f"\n💡 Interpretation: The mode of matches played is {matches_played_mode}. "
    f"This means most batters have played {matches_played_mode} matches."
)
````

````python
# 2.4 Coefficient of Variation (CV)

## Batters

# 1. Calculate CV for Runs
sd_runs = df_batters["Runs"].std()
mean_runs = df_batters["Runs"].mean()

cv_runs = (sd_runs / mean_runs) * 100

print(f"📊 Mean Runs: {mean_runs:.2f}")
print(f"📊 Standard Deviation of Runs: {sd_runs:.2f}")
print(f"📊 Coefficient of Variation (CV): {cv_runs:.2f}%")

print(
    f"\n💡Interpretation: The coefficient of variation (CV) for runs is "
    f"{cv_runs:.2f}%, meaning the runs vary by approximately {cv_runs:.2f}% "
    f"relative to the average runs scored."
)

if cv_runs < 20:
    consistency = "low variability and high consistency"
elif cv_runs < 50:
    consistency = "moderate variability"
else:
    consistency = "high variability and low consistency"

print(
    f"\n💡Interpretation: The CV for runs is {cv_runs:.2f}%, indicating "
    f"{consistency} in the runs scored by batters."
)
````

````python
# 2. Calculate CV for strike rates.
mean_sr = df_batters["SR"].mean()
std_sr = df_batters["SR"].std()
cv_sr = (std_sr / mean_sr) * 100

print(f"📊Mean Strike Rate: {mean_sr:.2f}")
print(f"📊SD of Strike Rate: {std_sr:.2f}")
print(f"📊CV of Strike Rate: {cv_sr:.2f}%")

print(
    f"\n💡Interpretation: The coefficient of variation (CV) for strike rate is {cv_sr:.2f}, means the strike rates vary by approximately {cv_sr:.2f} relative to the average strike rate"
)
````

````python
# 3. Which metric is more stable?
cv_sr = (df_batters["SR"].std() / df_batters["SR"].mean()) * 100
cv_avg = (df_batters["AVG"].std() / df_batters["AVG"].mean()) * 100
cv_runs = (df_batters["Runs"].std() / df_batters["Runs"].mean()) * 100


cv_dir = {"Runs": cv_runs, "SR": cv_sr, "AVG": cv_avg}

most_stable = min(cv_dir, key=cv_dir.get)

print(f"📊Most Stable Metric: {most_stable}")

print(
    f"💡Interpretation: {most_stable} is the most stable metric "
    f"because it has the lowest coefficient of variation, indicating "
    f"the least relative variability among batters."
)
````

````python
# 4. Compare CV of fours and sixes.
cv_four = (df_batters["4s"].std() / df_batters["4s"].mean()) * 100
cv_six = (df_batters["6s"].std() / df_batters["6s"].mean()) * 100

print(f"📊CV of Fours: {cv_four:.2f}%")
print(f"📊CV of Sixes: {cv_six:.2f}%")

if cv_four < cv_six:
    print(
        f"\n💡Interpretation: Fours are more consistent than sixes because "
        f"their CV ({cv_four:.2f}%) is lower than the CV of sixes ({cv_six:.2f}%)."
    )
elif cv_six < cv_four:
    print(
        f"\n💡Interpretation: Sixes are more consistent than fours because "
        f"their CV ({cv_six:.2f}%) is lower than the CV of fours ({cv_four:.2f}%)."
    )
else:
    print("\n💡Interpretation: Fours and sixes have the same relative variability.")
````

````python
# 5. Find team-wise CV of runs.
teams_group = df_batters.groupby("Team")
team_wise_cv = (
    (teams_group["Runs"].std() / teams_group["Runs"].mean()) * 100
).sort_values()

for team in team_wise_cv.index:
    print(f"📊CV of {team} : {team_wise_cv[team]:.2f}%")

lowest_cv_team = team_wise_cv.idxmin()
highest_cv_team = team_wise_cv.idxmax()

print(
    f"\n💡Interpretation: {lowest_cv_team} has the lowest CV "
    f"({team_wise_cv.min():.2f}%), indicating the most consistent run-scoring "
    f"performance among its batters."
)

print(
    f"💡{highest_cv_team} has the highest CV "
    f"({team_wise_cv.max():.2f}%), indicating the greatest variability "
    f"in runs scored by its batters."
)
````

````python
# 3. Which variable has more outliers?
def count_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = series[(series < lower_bound) | (series > upper_bound)]

    return len(outliers)


wkt_outliers = count_outliers(df_bowlers["WKT"])
avg_outliers = count_outliers(df_bowlers["AVG"])
eco_outliers = count_outliers(df_bowlers["ECO"])
sr_outliers = count_outliers(df_bowlers["SR"])

print(f"📊 WKT Outliers : {wkt_outliers}")
print(f"📊 AVG Outliers : {avg_outliers}")
print(f"📊 ECO Outliers : {eco_outliers}")
print(f"📊 SR Outliers  : {sr_outliers}")

outlier_counts = {
    "WKT": wkt_outliers,
    "AVG": avg_outliers,
    "ECO": eco_outliers,
    "SR": sr_outliers,
}

max_variable = max(outlier_counts, key=outlier_counts.get)

print(
    f"\n📊 Variable with most outliers: "
    f"{max_variable} ({outlier_counts[max_variable]})"
)
````

````python
# 4. Is economy left-skewed or right-skewed?
q1_eco = df_bowlers["AVG"].quantile(0.25)
median_eco = df_bowlers["AVG"].median()
q3_eco = df_bowlers["AVG"].quantile(0.75)

iqr_eco = q3_eco - q1_eco

upper_bound = q3_eco + 1.5 * iqr_eco
lower_bound = q1_eco - 1.5 * iqr_eco

lower_outliers_outliers = len(df_bowlers[df_bowlers["AVG"] < lower_bound])
upper_outliers = len(df_bowlers[df_bowlers["AVG"] > upper_bound])

print(f"📊 Lower Outliers: {lower_outliers}")
print(f"📊 Upper Outliers: {upper_outliers}")

if upper_outliers > lower_outliers:
    print(
        "\n💡Interpretation: The economy rate distribution is positively "
        "skewed because more outliers are present on the upper side."
    )
elif lower_outliers > upper_outliers:
    print(
        "\n💡Interpretation: The economy rate distribution is negatively "
        "skewed because more outliers are present on the lower side."
    )
else:
    print("\n💡Interpretation: The distribution appears fairly symmetric.")
````

````python
# 5. Find covariance matrix for numeric columns.

from itertools import combinations

# Numeric columns
cols = ["Runs", "SR", "BF", "AVG"]

# Covariance Matrix
cov_matrix = df_batters[cols].cov()

print("📊 Covariance Matrix")
print(cov_matrix)

print("\n" + "=" * 60)
print("📊 Pairwise Covariance Analysis")
print("=" * 60)

# Pairwise covariance
for col1, col2 in combinations(cols, 2):

    cov_value = cov_matrix.loc[col1, col2]

    print(f"\n📊 Covariance between {col1} and {col2}: {cov_value:.2f}")

    if cov_value > 0:
        print(
            f"💡 Positive Covariance: Higher {col1} tends to be associated with higher {col2}"
        )

    elif cov_value < 0:
        print(
            f"💡 Negative Covariance: Higher {col1} tends to be associated with lower {col2}"
        )

    else:
        print(f"💡 No Covariance: No linear relationship between {col1} and {col2}")
````
