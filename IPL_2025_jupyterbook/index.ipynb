````python
# PART 1: Measures of Central Tendency

# 1.1 Mean (Average)

## Batters Dataset Questions

# 1. Find the average (mean) runs scored by all batters

batters_avg = df_batters["Runs"].mean()

print(f"📊 Mean runs scored by all batters: {batters_avg:.2f}")

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(df_batters["Runs"], alpha=0.7, edgecolor="black")

# Mean Line
plt.axvline(
    batters_avg,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean = {batters_avg:.2f}",
)

# Labels and Title
plt.xlabel("Runs Scored")
plt.ylabel("Frequency")
plt.title("Distribution of Runs Scored by Batters")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# Interpretation
print(
    f"\n💡 Interpretation: "
    f"The average runs scored by batters in IPL 2025 is approximately {batters_avg:.2f}."
)
````

````python
# 2. Calculate mean strike rate (SR) of all batters.
batters_sr = df_batters["SR"].mean()
print(f"\n📊 Mean Strike Rate by all batters: {batters_sr:.2f}")

# Visualisation
plt.figure(figsize=(10, 5))
plt.hist(df_batters["SR"], edgecolor="black")
plt.axvline(
    batters_sr,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean Strike Rate = {batters_sr:.2f}",
)

plt.xlabel("Strike Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Strike Rate for All Batters")
plt.grid(True, alpha=0.3)

plt.legend()
plt.show()

print(
    f"\n💡 Interpretation : The average strike rate for Batters in IPL 2025 was approximately {batters_sr:.2f}"
)
````

````python
# 3. Find average number of sixes hit by batters.
sixes_hit = df_batters["6s"].mean()
print(f"📊 Mean Sixes hit by all batters : {sixes_hit:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_batters["6s"], edgecolor="black", alpha=0.7)
plt.axvline(
    sixes_hit, linestyle="--", linewidth=2, label=f"Mean sixes hit {sixes_hit:.2f}"
)
plt.legend()
plt.xlabel("Sixes Hit By Batter")
plt.ylabel("Frequency")
plt.title("Distribution of Sixes Hit By Batters")
plt.grid(alpha=0.3)
plt.show()

print(
    f"💡 Interpretation : The average Sixes Hit by the Batters in IPL 2025 is {sixes_hit:.2f}"
)
````

````python
# 4. Which batter scored closest to the average runs?
# Average Runs
average_runs = df_batters["Runs"].mean()

# Closest Batter Index
closest_index = (df_batters["Runs"] - average_runs).abs().idxmin()

# Closest Batter Data
closest_batter = df_batters.loc[closest_index]

# Values
closest_runs = closest_batter["Runs"]
closest_name = closest_batter["Player Name"]

# Plot
plt.figure(figsize=(10, 5))

# Histogram
plt.hist(df_batters["Runs"], bins=10, edgecolor="black", alpha=0.7)

# Mean Line
plt.axvline(
    average_runs,
    color="red",
    linestyle="--",
    linewidth=2,
    label=f"Mean Runs = {average_runs:.2f}",
)

# Closest Batter Line
plt.axvline(
    closest_runs,
    color="green",
    linestyle="-",
    linewidth=2,
    label=f"{closest_name} = {closest_runs}",
)

# Labels
plt.xlabel("Runs")
plt.ylabel("Frequency")
plt.title("Runs Distribution with Closest Batter to Average")

plt.legend()

plt.grid(True, alpha=0.3)
plt.show()
````

````python
# 5. Find mean batting average (AVG) team-wise.
# Team Color Mapping
team_colors = {
    "MI": "#004BA0",
    "CSK": "#F9CD05",
    "RCB": "#D71920",
    "KKR": "#3A225D",
    "SRH": "#FF822A",
    "DC": "#17449B",
    "RR": "#EA1A85",
    "PBKS": "#D71920",
    "LSG": "#00AEEF",
    "GT": "#1B2133",
}


# Convert AVG column to numeric
df_batters["AVG"] = pd.to_numeric(df_batters["AVG"], errors="coerce")

# Team-wise batting average
team_wise_avg = df_batters.groupby("Team")["AVG"].mean().sort_values(ascending=False)

print("\n📊 Team-wise Mean Batting Average:\n")
print(team_wise_avg)

colors = [team_colors[team] for team in team_wise_avg.index]

# Visualization
plt.figure(figsize=(10, 5))

ax=team_wise_avg.plot(kind="bar", edgecolor="black", alpha=0.7,color=colors)

plt.xlabel("Team")
plt.ylabel("Mean Batting Average")
plt.title("Team-wise Mean Batting Average")
plt.bar_label(
ax.containers[0],
fmt='%.2f'
)

plt.grid(True, alpha=0.3)

plt.show()

# Interpretation
print(
    "\n💡 Interpretation: "
    "This graph shows the average batting performance of players across different IPL teams."
)
````

````python
## Bowlers Dataset Questions

# 1. Find average wickets (WKT) taken by bowlers.
avg_wickets = df_bowlers["WKT"].mean()
print(f"\n📊 Average wickets (WKT) taken by bowlers {avg_wickets:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_bowlers["WKT"], alpha=0.7, edgecolor="black")
plt.axvline(
    avg_wickets,
    linestyle="--",
    color="red",
    linewidth=2,
    label=f"Mean Wickets Taken : {avg_wickets:.2f}",
)
plt.xlabel("Wickets Taken by Bowlers")
plt.ylabel("Frequency")
plt.title("Distribution of Wickets taken by bowlers")

plt.legend()

plt.show()

print(
    f"\n💡 Interpretation: The average Wickets taken by the players in IPL 2025 is {avg_wickets}"
)
````

````python
# 2. Calculate mean economy rate (ECO).

avg_economy = df_bowlers["ECO"].mean()
print(f"\n📊 Mean economy rate by the bowlers {avg_economy:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_bowlers["ECO"], alpha=0.7, edgecolor="black")
plt.axvline(
    avg_economy,
    linestyle="--",
    linewidth=2,
    color="red",
    label=f"Mean Economy Rate {avg_economy:.2f}",
)
plt.xlabel("Economy Rate (ECO)")
plt.ylabel("Frequency")
plt.title("Distribution of Economy Rate")

plt.legend()

plt.show()

print(
    f"\n 💡 Interpretation : The average economy rate for IPL 2025 is {avg_economy:.2f}"
)
````

````python
# 3. Find average bowling strike rate (SR).
mean_strikerate = df_bowlers["SR"].mean()
print(f"\n📊 Mean Bowling Strike Rate by all the bowlers {mean_strikerate:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_bowlers["SR"], bins=10, alpha=0.7, edgecolor="black")
plt.axvline(
    mean_strikerate,
    linestyle="--",
    linewidth=2,
    color="red",
    label=f"Mean Strike Rate {mean_strikerate:.2f}",
)

plt.xlabel("Bowlers Strike Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Bowlers Strike Rate")
plt.legend()
plt.show()

print(
    f"\n💡 Interpretation: Mean Bowling Strike Rate by all the bowlers in IPL 2025 is {mean_strikerate:.2f}"
)
````

````python
# 4. Which bowler has wickets closest to mean wickets?
mean_wickets = df_bowlers["WKT"].mean()
closest_bowler_index = (df_bowlers["WKT"] - mean_wickets).abs().idxmin()
closest_bowler = df_bowlers.loc[closest_bowler_index]

# value
bowler_name = closest_bowler["Player Name"]
bowler_wkt = closest_bowler["WKT"]

print(
    f"\n📊 The closest bowler to mean wickets is {bowler_name} with {bowler_wkt} wickets"
)

plt.figure(figsize=(10, 5))
plt.hist(df_bowlers["WKT"], alpha=0.7, bins=10, edgecolor="black")

plt.axvline(
    mean_wickets,
    linestyle="--",
    color="red",
    linewidth=2,
    label=f"Mean Wickets = {mean_wickets}",
)
plt.axvline(
    bowler_wkt,
    linestyle="-",
    color="green",
    linewidth=2,
    label=f"{bowler_name} = {bowler_wkt}",
)
plt.xlabel("Wickets taken by bowlers")
plt.ylabel("Frequency")
plt.title("Distribution of Wickets by Bowlers")

plt.legend()
plt.show()

print(
    f"💡 The closest bowler to mean wickets is {bowler_name} with {bowler_wkt} wickets"
)
````

````python
# 5. Find average wickets team-wise.

# Team Color Mapping
team_colors = {
    "MI": "#004BA0",
    "CSK": "#F9CD05",
    "RCB": "#D71920",
    "KKR": "#3A225D",
    "SRH": "#FF822A",
    "DC": "#17449B",
    "RR": "#EA1A85",
    "PBKS": "#D71920",
    "LSG": "#00AEEF",
    "GT": "#1B2133",
}

# Team-wise Average Wickets
average_wkt = df_bowlers.groupby("Team")["WKT"].mean().sort_values(ascending=True)
display(average_wkt.index)

# Auto Colors
colors = [team_colors[team] for team in average_wkt.index]

# Plot
plt.figure(figsize=(10, 5))

ax = average_wkt.plot(kind="bar", color=colors, alpha=0.7, edgecolor="black")
ax.bar_label(ax.containers[0], fmt="%.2f")


plt.xlabel("Team")
plt.ylabel("Average Wickets")
plt.title("Team-wise Average Wickets")

plt.grid(True, alpha=0.3)

plt.show()
````

````python
# 1.2 Median

## Batters

# 1. Find median runs scored.
median_runs_scored = df_batters["Runs"].median()

print(f"📊 Median Runs Scored: {median_runs_scored:.2f}")

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(df_batters["Runs"], alpha=0.7, edgecolor="black", bins=10)

# Median Line
plt.axvline(
    median_runs_scored,
    linestyle="--",
    linewidth=2,
    color="blue",
    label=f"Median Runs = {median_runs_scored:.2f}",
)

# Labels and Title
plt.xlabel("Runs Scored")
plt.ylabel("Frequency")
plt.title("Distribution of Runs Scored")

plt.legend()

plt.grid(True, alpha=0.3)

plt.show()

# Interpretation
print(
    f"\n💡 Interpretation: "
    f"\nThe median runs scored by batters is {median_runs_scored:.2f}. "
    f"\nThis means 50% of batters scored below this value and 50% scored above it."
)
````

````python
# 2. Find median strike rate.
median_strike_rate = df_batters["SR"].median()
print(f"📊 Median Strike Rate: {median_strike_rate:.2f}")

plt.figure(figsize=(10, 5))
plt.hist(df_batters["SR"], bins=10, alpha=0.7, edgecolor="black")
plt.axvline(
    median_strike_rate,
    linestyle="--",
    linewidth=2,
    color="blue",
    label=f"Median Strike Rate = {median_strike_rate:.2f}",
)
plt.xlabel("Strike Rate")
plt.ylabel("Frequency")
plt.title("Distribution of Batters Strike Rate")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print(f"\n💡 Interpretation: The median Strike Rate by batter is {median_strike_rate}.")
print(
    f"\nThis means 50% of batters have a strike rate below this value and 50% have a strike rate above it."
)
````

````python
# 3. Find median number of fours.
median_fours = df_batters["4s"].median()
print(f"Median number of fours : {median_fours}")

plt.figure(figsize=(10,5))

plt.hist(df_batters['4s'],bins=10,alpha=.7,edgecolor='black')
plt.axvline(median_fours,linestyle='--',linewidth=2,color="blue",label=f"Media Fours = {median_fours:.2f}")
plt.legend()
plt.xlabel('Fours')
plt.ylabel("Frequency")
plt.title("Distribution of number of fours")
plt.grid(True,alpha=.3)
plt.show()

print(f"\n💡Interpretation: The median number of fours hit by batsmen is {median_fours}")
print(f"This means 50% of batters have hit fours below this value and 50% of batters have hit above.")
````

````python
# 4. Compare mean vs median runs. Is the data skewed?

# Median
median_runs = df_batters["Runs"].median()
print(f"📊 Median Runs Scored: {median_runs:.2f}")

# Mean
mean_runs = df_batters["Runs"].mean()
print(f"📊 Mean Runs Scored: {mean_runs:.2f}")

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(df_batters["Runs"], bins=10, alpha=0.7, edgecolor="black")

# Median Line
plt.axvline(
    median_runs,
    linestyle="--",
    linewidth=2,
    color="blue",
    label=f"Median = {median_runs:.2f}",
)

# Mean Line
plt.axvline(
    mean_runs, linestyle="-", linewidth=2, color="red", label=f"Mean = {mean_runs:.2f}"
)

# Labels and Title
plt.xlabel("Runs Scored")
plt.ylabel("Frequency")
plt.title("Distribution of Runs Scored")

plt.grid(True, alpha=0.3)

plt.legend()

plt.show()

# Skewness Interpretation
if mean_runs > median_runs:
    print(
        "\n💡 Interpretation: "
        "The data is right skewed because the mean is greater than the median. "
        "This indicates the presence of a few high-scoring batters pulling the average upward."
    )

elif mean_runs < median_runs:
    print(
        "\n💡 Interpretation: "
        "The data is left skewed because the mean is smaller than the median."
    )

else:
    print(
        "\n💡 Interpretation: "
        "The data is approximately symmetric because the mean and median are nearly equal."
    )
````

````python
# 5. Find median batting average for batters with more than 500 runs

# Convert AVG column to numeric
df_batters["AVG"] = pd.to_numeric(df_batters["AVG"], errors="coerce")

# Filter batters with more than 500 runs
high_run_batters = df_batters[df_batters["Runs"] > 500]

# Median AVG
avg_median_data = high_run_batters["AVG"].median()

print(
    f"📊 Median Batting Average for batters "
    f"with more than 500 runs: {avg_median_data:.2f}"
)

# Visualization
plt.figure(figsize=(10, 5))

plt.hist(high_run_batters["AVG"], alpha=0.7, edgecolor="black", bins=10)

# Median Line
plt.axvline(
    avg_median_data,
    linestyle="--",
    linewidth=2,
    color="red",
    alpha=0.8,
    label=f"Median AVG = {avg_median_data:.2f}",
)

# Labels and Title
plt.xlabel("Batting Average")
plt.ylabel("Frequency")
plt.title("Distribution of Batting Average " "for Batters with 500+ Runs")

plt.grid(True, alpha=0.3)

plt.legend()

plt.show()

# Interpretation
print(
    f"\n💡 Interpretation: "
    f"The median batting average for batters with "
    f"more than 500 runs is {avg_median_data:.2f}. "
    f"This represents the typical batting average "
    f"among high-performing batters."
)
````
